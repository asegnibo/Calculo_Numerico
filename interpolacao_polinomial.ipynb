```python
dados['Mês'][0]

def lagrange_interpolation(x_points, y_points, x):
    """
    Calcula o valor interpolado usando o método de Lagrange.

    Args:
        x_points: Uma lista ou array numpy de coordenadas x dos pontos de dados.
        y_points: Uma lista ou array numpy de coordenadas y dos pontos de dados.
        x: O ponto no qual se deseja calcular a interpolação.

    Returns:
        O valor interpolado em x.
    """
    n = len(x_points)
    if n != len(y_points):
        raise ValueError("x_points e y_points devem ter o mesmo comprimento")

    interpolated_value = 0
    for i in range(n):
        term = y_points[i]
        for j in range(n):
            if i != j:
                term = term * (x - x_points[j]) / (x_points[i] - x_points[j])
        interpolated_value += term
    return interpolated_value
```
